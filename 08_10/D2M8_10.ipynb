```python
#Tuples
student=("Yathees",25,"CS")
print(student)
print(student[0])
for detail in student:
    print(detail)

#lIST - MUTABLE (can change)
#TUPLE - IMMUTABLE
```

```python
#Dictionary
student={
    "Name":"Yathees",
    "Age":25,
    "Field":"CS"
}
print(student)
print(student["Name"])
print("Age",student["Age"])
for value in student:
    print( value)
```

```python
#function
def calc_avg(a,b):
    return (a+b)/2

result=calc_avg(10,20)
print(result)
```

```python
def add(a,b):
    return(a+b)

def mul(a,b):
    return(a*b)

def div(a,b):
    return(a/b)

def sub(a,b):
    return(a-b)

print("a+b=",add(2,4))
print("a*b=",mul(2,4))
print("a/b=",div(20,4))
print("a-b=",sub(2,4))
```
