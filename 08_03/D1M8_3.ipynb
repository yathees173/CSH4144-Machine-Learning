```python
marks=75

#IF Statement
if marks>=50:
    print("1_Pass")

#IF-ELSE Statements
if marks>=50:
    print("2_Pass")

else:
    print("Fail")
    
#IF-ELIF-ELSE Statement
if marks>=75:
    print("Grade A")
        
elif marks>=65:
    print("Grade B")
else:
    print("Fail")
    
```
