```python
benign_df=df[df['Class']==2][0:200]
malignant_df=df[df['Class']==4][0:200]
axes=benign_df.plot(kind='scatter',
                   x='Clump', y='UnifSize',
                   color='blue', label='Beningn')
malignant_df.plot(kind='scatter', 
                 x='Clump', y='UnifSize',
                 color='red', label='Malignant')
```

```python
df.dtypes
```

```python
df=df[pd.to_numeric(df['BareNuc'],
                   errors='coerce').notnull()]
df['BareNuc']=df['BareNuc'].astype('int')
```

```python
df.dtypes
```

```python
df.columns
```

```python
#Select the input features
feature_df=df[['Clump', 'UnifSize', 'UnifShape', 'MargAdh', 'SingEpiSize',
       'BareNuc', 'BlandChrom', 'NormNucl', 'Mit']]
```

```python
#Convert the features into a Numpy array
X=np.asarray(feature_df)
```

```python
#Select the target variable
y=np.asarray(df['Class'])
```

```python
#Displaythe first five rows
X[0:5]
```

```python
#Split the data
X_train, X_test, y_train, y_test= train_test_split(X,y,test_size=0.2, random_state=1)
```

```python
X_train.shape
```

```python
from sklearn import svm
#support vector classifier
```

```python
#Train the classifier
classifier=svm.SVC(kernel='linear', gamma='auto', C=2)
classifier.fit(X_train,y_train)
```

```python
#Predict the classes of the test data
y_predict=classifier.predict(X_test)
```

```python
#import the classification report
from sklearn.metrics import classification_report

#print the classification report
print(classification_report(y_test,y_predict))
```
